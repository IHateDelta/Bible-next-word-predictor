```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import keras as kr
import seaborn as sb
import sklearn
import tensorflow as tf
from keras.preprocessing.sequence import pad_sequences
from keras import layers, Model
from sklearn.metrics import accuracy_score

sklearn.set_config(transform_output="pandas")
```

```python
in_file=open("bible_ready.txt","r", errors='ignore',encoding='utf8')
text=in_file.read()
in_file.close()
T=np.array(text.split("\n"))
ord_len=len(T)
print(T)
```

```python
WS=np.array(sorted(set(T)))
ws_len=len(WS)+1
word_to_id = {word: i+1 for i, word in enumerate(WS)}
id_to_word = {i+1: word for i, word in enumerate(WS)}
print(id_to_word)
```

```python
ordered_ids = [word_to_id[w] for w in T]
ordered_ids
```

```python
SEQ_CHUNK = 10
embed_dim = 100
sequences = []
for i in range(0, len(ordered_ids), SEQ_CHUNK):
    chunk = ordered_ids[i : i + SEQ_CHUNK]
    if len(chunk) > 1:
        sequences.append(chunk)
X_list = [seq[:-1] for seq in sequences]
y_list = [seq[1:]  for seq in sequences]
X = pad_sequences(X_list, padding='post', value=0)
y = pad_sequences(y_list, padding='post', value=0)
len(X_list)
```

```python
"""
text_prep lines.py"model = kr.Sequential([
        kr.layers.Input(shape=(None,)),
        kr.layers.Embedding(input_dim=ws_len, output_dim=embed_dim, mask_zero=True),
        kr.layers.LSTM(embed_dim, return_sequences=True, name='lstm_layer'),
        kr.layers.Dense(ws_len,activation="sigmoid"),
        #kr.layers.Embedding(input_dim=embed_dim, output_dim=ws_len, mask_zero=True),
        kr.layers.Activation(activation="softmax")
    ])
"""
```

```python
input_layer_1 = layers.Input(shape=(None,), dtype='int32', name='input_ids')
embedding_layer_2 = layers.Embedding(input_dim=ws_len, output_dim=embed_dim, mask_zero=True)
lstm_layer_3 = layers.LSTM(embed_dim, return_sequences=True)
lstm_layer_4 = layers.LSTM(embed_dim, return_sequences=True,)
lstm_layer_5 = layers.LSTM(embed_dim, return_sequences=True)
dense_layer_6=layers.Dense(200,activation="sigmoid")
dense_layer_7=layers.Dense(200,activation="sigmoid")
dense_layer_8=layers.Dense(100,activation="sigmoid")
def tied_dense(x):
    return tf.matmul(x, embedding_layer_2.embeddings, transpose_b=True)
def propagate_mask(inputs, mask=None):
    return mask
reverse_embeding_layer_9=layers.Lambda(tied_dense,mask=propagate_mask)
output_layer_10 = layers.Activation('softmax')

input=input_layer_1
h1=embedding_layer_2(input)
h2=lstm_layer_3(h1)
h3=lstm_layer_4(h2)
h4=lstm_layer_5(h3)
h5=dense_layer_6(h4)
h6=dense_layer_7(h5)
h7=dense_layer_8(h6)

h8=reverse_embeding_layer_9(h7)
output=output_layer_10(h8)
```

```python
model=Model(inputs=input,outputs=output)
model.summary()
```

```python
Adam = kr.optimizers.Adam(learning_rate=0.1)
#r_lr=kr.callbacks.ReduceLROnPlateau(patience=5, monitor="val_loss",factor=0.4)
#e_s=kr.callbacks.EarlyStopping(patience=12, monitor="val_loss")
```

```python
model.compile(optimizer=Adam,loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
history = model.fit(X, y,epochs=3,validation_split=0.01)
```

```python
model.save('smallsec_3.keras')
```

```python
def next_word(ws_test):
    seq=np.array([ws_test])
    prediction=model.predict(seq)
    last_word_proba=prediction[0, -1, :]
    #next_word_id = np.argmax(last_word_proba)
    next_word_id = np.random.choice(len(last_word_proba), p=last_word_proba)
    print(id_to_word[next_word_id],last_word_proba[next_word_id])
    return next_word_id
def generate_me_bible(S,n):
    Tab=np.array(S.split(" "))
    ws_test=[word_to_id[w] for w in Tab]
    for i in range(n):
        next=next_word(ws_test)
        ws_test.append(next)
    result=""
    for w in ws_test:
        result+=id_to_word[w]+" "
    return result
```
